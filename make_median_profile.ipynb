```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
```

```python
file = 'scidar_lapalma_cn2dhs_2020A.fits'
#file = 'scidar_paranal_cn2dhs_2019Abeta_conj1860 2.fits'
hdul = fits.open(file)
data = hdul[0].data
height = h = np.arange(0,25000,250)
```

```python
data[data<=0] = np.nan
median_profile = np.nanmedian(data,axis=0)
sumCn2dh = np.nansum(data,axis=1)
r0 = (0.423*(2.*np.pi/500e-9)**2*sumCn2dh)**(-3./5.)
np.median(r0)
```

```python
plt.plot(median_profile,height,label='median',color='r')
for i in np.arange(data.shape[0]):
    plt.plot(data[i],height,color='b',alpha=0.005)
plt.xscale('log')
plt.xlim(1e-15/250,1e-11)
plt.show()
```

```python
export = [height,median_profile]
np.savetxt('la-palma-median.csv',export,delimiter=',')
```
